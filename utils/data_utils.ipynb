```python
# pobranie złotych danych 

import pandas as pd

URL = "https://freegoldapi.com/data/latest.csv"  

df = pd.read_csv(URL)

df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"])

df["yearmonth"] = df["date"].dt.to_period("M").astype(str)

monthly = (
    df.groupby("yearmonth", as_index=False)["price"]
      .mean()
      .rename(columns={"yearmonth": "date", "price": "price_usd_per_troy_oz"})
)

monthly = monthly[(monthly["date"] >= "2005-01") & (monthly["date"] <= "2025-12")]

monthly.to_csv("gold_price_monthly_USD_2005_2025.csv", index=False)
print("Zapisano: gold_price_monthly_USD_2005_2025.csv")
```

```python
# pobranie srebrnych danych
import pandas as pd

URL = "https://stooq.pl/q/d/l/?s=xagusd&i=d"
df = pd.read_csv(URL)

df["date"] = pd.to_datetime(df["Data"], errors="coerce")
df["price"] = pd.to_numeric(df["Zamkniecie"], errors="coerce")
df = df.dropna(subset=["date","price"])

df["yearmonth"] = df["date"].dt.to_period("M").astype(str)

monthly = (
    df.groupby("yearmonth", as_index=False)["price"]
      .mean().round(3)
      .rename(columns={"yearmonth": "date", "price": "price_usd_per_troy_oz"})
)

monthly = monthly[(monthly["date"] >= "2005-01") & (monthly["date"] <= "2025-12")]

monthly.to_csv('silver_price_monthly_USD_2005_2025.csv', index=False)
print('Zapisano: silver_price_monthly_USD_2005_2025.csv')
```
